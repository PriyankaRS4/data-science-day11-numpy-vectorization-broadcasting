# Task 11 — NumPy Vectorization and Broadcasting

**Objective:** Perform arithmetic operations on complete NumPy arrays, demonstrate broadcasting with compatible shapes, and compare loop-based vs vectorized calculations.

Dataset used: **Iris dataset** (`iris.csv`)


```python
import numpy as np
from pathlib import Path
import time

# Load the 4 numeric columns from the Iris CSV
csv_path = Path("iris.csv")
if not csv_path.exists():
    csv_path = Path("/mnt/data/iris.csv")  # fallback for this environment

X = np.loadtxt(csv_path, delimiter=",", skiprows=1, usecols=(0, 1, 2, 3))

print("Shape of numeric Iris data:", X.shape)
print("\nFirst 5 rows:")
print(X[:5])
```

## 1. Vectorized Arithmetic

```python
# Arithmetic on complete arrays — no explicit Python loops
print("Original first 3 rows:\n", X[:3])

print("\nAdd 2:")
print(X[:3] + 2)

print("\nSubtract 1:")
print(X[:3] - 1)

print("\nMultiply by 10:")
print(X[:3] * 10)

print("\nDivide by 2:")
print(X[:3] / 2)

print("\nSquare every value:")
print(X[:3] ** 2)
```

## 2. Broadcasting Examples

```python
# Broadcasting Example 1: scalar + array
# Scalar 1.0 is broadcast to every element of the array.
example1 = X[:3] + 1.0

print("Example 1 — Scalar broadcasting:")
print("X[:3] shape:", X[:3].shape)
print("Result:\n", example1)
```

```python
# Broadcasting Example 2: 1D feature vector + 2D array
# Shape (4,) is broadcast across all rows.
feature_adjustment = np.array([0.1, 0.2, 0.3, 0.4])
example2 = X[:3] + feature_adjustment

print("Example 2 — Row/feature-vector broadcasting:")
print("Data shape:", X[:3].shape)
print("Adjustment shape:", feature_adjustment.shape)
print("Result:\n", example2)
```

```python
# Broadcasting Example 3: column vector with a 2D array
# Each row is divided by its own maximum value.
row_max = X[:5].max(axis=1, keepdims=True)   # shape (5, 1)
example3 = X[:5] / row_max

print("Example 3 — Column-vector broadcasting:")
print("Data shape:", X[:5].shape)
print("row_max shape:", row_max.shape)
print("Result:\n", example3)
```

## 3. Useful Vectorized Data-Science Operation

```python
# Center each feature by subtracting its column mean.
# Column mean has shape (4,) and is broadcast across all 150 rows.
column_means = X.mean(axis=0)
centered = X - column_means

print("Column means:", column_means)
print("\nFirst 3 centered rows:")
print(centered[:3])
```

## 4. Loop-Based vs Vectorized Calculation

```python
# Use a larger repeated array so the speed difference is easy to observe.
big_X = np.tile(X, (1000, 1))

# Loop-based square
start = time.perf_counter()
loop_result = np.empty_like(big_X)

for i in range(big_X.shape[0]):
    for j in range(big_X.shape[1]):
        loop_result[i, j] = big_X[i, j] ** 2

loop_time = time.perf_counter() - start

# Vectorized square
start = time.perf_counter()
vector_result = big_X ** 2
vector_time = time.perf_counter() - start

print(f"Loop time:       {loop_time:.6f} seconds")
print(f"Vectorized time: {vector_time:.6f} seconds")
print(f"Speed-up:        {loop_time / vector_time:.2f}x")
print("Same result:", np.allclose(loop_result, vector_result))
```

## Conclusion

- **Vectorization** performs operations on whole NumPy arrays instead of writing explicit Python loops.
- **Broadcasting** allows NumPy to operate on arrays with compatible shapes.
- Vectorized NumPy code is usually **shorter, cleaner, and much faster** because the heavy computation is performed in optimized compiled code.


## Interview Questions

**1. What is vectorization?**  
Vectorization means applying an operation to an entire NumPy array at once instead of processing elements one by one using Python loops.

**2. What is broadcasting in NumPy?**  
Broadcasting is NumPy's ability to perform arithmetic between arrays of different but compatible shapes by automatically expanding the smaller shape conceptually.

**3. Why is vectorization useful for data science?**  
It improves execution speed, reduces Python code, makes calculations easier to read, and works efficiently with large numerical datasets.
